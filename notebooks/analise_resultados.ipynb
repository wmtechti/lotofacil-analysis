# 🎯 MEGA-SENA - Análise Geométrica (Grid 6×10)

Grid de 60 números dispostos em 6 linhas × 10 colunas, formando padrões tipo "Batalha Naval".

## 2️⃣ Carregar Dados da Mega-Sena

````python
# Carregar análises da Mega-Sena
with open('../out/megasena/megasena_analyses.json', 'r', encoding='utf-8') as f:
    megasena_analyses = json.load(f)

print(f"📊 Total de sorteios analisados: {len(megasena_analyses)}")
print(f"📅 Período: Concurso {megasena_analyses[0]['concurso']} até {megasena_analyses[-1]['concurso']}")

# Criar DataFrame resumido
megasena_df = pd.DataFrame([
    {
        'Concurso': a['concurso'],
        'Números': str(sorted(a['numeros'])),
        'Padrão': a['pattern'],
        'Pares_Conectados': a['contiguity']['connected_pairs'],
        'Componentes': a['contiguity']['connected_components'],
        'Adjacência_%': a['contiguity']['adjacency_ratio'] * 100,
        'Dispersão_Média': a['dispersion']['mean_pairwise_distance'],
        'Dispersão_Máxima': a['dispersion']['max_distance']
    }
    for a in megasena_analyses
])

# Exibir primeiros registros
megasena_df.head(10)
````

## 3️⃣ Estatísticas Gerais - Mega-Sena

````python
# Estatísticas descritivas
print("=" * 80)
print("📊 ESTATÍSTICAS DESCRITIVAS - MEGA-SENA")
print("=" * 80)

print("\n🔢 Variáveis Numéricas:")
print(megasena_df[['Pares_Conectados', 'Componentes', 'Adjacência_%', 
                    'Dispersão_Média', 'Dispersão_Máxima']].describe())

print("\n📐 Distribuição de Padrões Geométricos:")
pattern_counts = megasena_df['Padrão'].value_counts()
for pattern, count in pattern_counts.items():
    percentage = 100 * count / len(megasena_df)
    print(f"  {pattern:20s}: {count:4d} sorteios ({percentage:5.2f}%)")

print("\n🏆 Top 10 Sorteios Mais Conectados:")
top_connected = megasena_df.nlargest(10, 'Pares_Conectados')[['Concurso', 'Números', 'Pares_Conectados', 'Padrão']]
print(top_connected.to_string(index=False))
````

### 4.6 Exemplos de Sorteios com Padrões

````python
# Exibir primeiros 6 sorteios em grid 2×3
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()

for i in range(6):
    img_path = f'../out/megasena/draw_examples/concurso_{i+1:04d}.png'
    if Path(img_path).exists():
        img = Image.open(img_path)
        axes[i].imshow(img)
        axes[i].axis('off')
    else:
        axes[i].text(0.5, 0.5, 'Imagem não encontrada', ha='center', va='center')
        axes[i].axis('off')

plt.suptitle('Exemplos de Sorteios - Padrões Geométricos', fontsize=18, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
````
