## 📌 Conclusões

### 🔍 Principais Insights:

1. **Padrão Dominante: DISPERSO (78.13%)**
   - A maioria dos sorteios forma padrões extremamente espalhados
   - Números tendem a ocupar áreas distantes do grid

2. **Baixíssima Contiguidade**
   - Média de apenas 0.87 pares adjacentes por sorteio
   - 38.69% dos sorteios são totalmente dispersos (sem números vizinhos)
   - **Nenhum sorteio formou figura totalmente conectada**

3. **Equilíbrio Regional Perfeito**
   - Distribuição uniforme entre quadrantes (~25% cada)
   - Não há viés para nenhuma região do grid

4. **Alta Dispersão Espacial**
   - Distância média entre números: 5.34 unidades
   - Números tendem a se espalhar por todo o grid
   - Variação significativa entre sorteios (2.53 a 7.87)

```python
# Carregar análises da Mega-Sena
with open('../out/megasena/megasena_analyses.json', 'r', encoding='utf-8') as f:
    megasena_analyses = json.load(f)

print(f"📊 Total de sorteios analisados: {len(megasena_analyses)}")
print(f"📅 Período: Concurso {megasena_analyses[0]['concurso']} até {megasena_analyses[-1]['concurso']}")

# Criar DataFrame resumido
megasena_df = pd.DataFrame([
    {
        'Concurso': a['concurso'],
        'Números': str(sorted(a['numeros'])),
        'Padrão': a['pattern'],
        'Pares_Conectados': a['contiguity']['connected_pairs'],
        'Componentes': a['contiguity']['connected_components'],
        'Adjacência_%': a['contiguity']['adjacency_ratio'] * 100,
        'Dispersão_Média': a['dispersion']['mean_pairwise_distance'],
        'Dispersão_Máxima': a['dispersion']['max_distance']
    }
    for a in megasena_analyses
])

# Exibir primeiros registros
megasena_df.head(10)
```
