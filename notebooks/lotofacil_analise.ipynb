## 📌 Conclusões e Insights

### 🔍 Principais Descobertas:

#### 1. **Frequências**
- Coluna 5 é a mais frequente (10.948 aparições)
- Linha 3 é a mais frequente (10.839 aparições)
- Distribuição relativamente equilibrada entre células

#### 2. **Co-ocorrência**
- **43 super pares** identificados (>95% força)
- Par mais forte: **[11+20]** com 1.362 aparições
- Pares fortes tendem a aparecer em ~38% dos sorteios

#### 3. **Números Quentes e Frios**
- Número mais quente: **20** (+4.06% acima da média)
- Número mais frio: **16** (-4.62% abaixo da média)
- Desvio máximo: ~4.6% (variação pequena confirma aleatoriedade)

#### 4. **Tendências Temporais (últimos 500 sorteios)**
- Número em maior alta: **20** (+8.14%)
- Número em maior baixa: **11** (-6.12%)
- Tendências são voláteis e podem reverter

#### 5. **Simulação Monte Carlo**
- **10.000 simulações** executadas
- Taxa de prêmio aleatório: **11.34%**
- Melhor estratégia: **Tendência Alta** (12.42%)
- Ganho sobre aleatório: **+1.08%**

### 💡 Recomendações:

✅ **Estratégia "Tendência Alta"** mostrou melhor performance
✅ Considerar **super pares** na seleção de números
✅ Equilibrar números de diferentes **regiões do grid**
✅ Não confiar em padrões de curto prazo (alta volatilidade)
✅ Diversificar apostas usando múltiplas estratégias

### ⚠️ Avisos Importantes:

- Nenhuma estratégia garante vitória
- Resultados passados não garantem resultados futuros
- Melhor estratégia teve apenas +1.08% vs aleatório
- Jogue com responsabilidade!

## 8️⃣ Resultados da Simulação Monte Carlo

```python
tendencias = pd.read_csv('../out/lotofacil/tendencias_temporais.csv')

# Top números em alta
em_alta = tendencias.nlargest(10, 'tendencia_percentual')
print("📈 TOP 10 NÚMEROS EM ALTA (últimos 500 sorteios):")
print(em_alta[['numero', 'freq_recente', 'freq_geral', 'tendencia_percentual']].to_string(index=False))

# Top números em baixa
em_baixa = tendencias.nsmallest(10, 'tendencia_percentual')
print("\n\n📉 TOP 10 NÚMEROS EM BAIXA:")
print(em_baixa[['numero', 'freq_recente', 'freq_geral', 'tendencia_percentual']].to_string(index=False))
```

## 7️⃣ Análise de Tendências Temporais

```python
quentes_frios = pd.read_csv('../out/lotofacil/numeros_quentes_frios.csv')

print("🔥 TOP 10 NÚMEROS MAIS QUENTES:")
print(quentes_frios.head(10).to_string(index=False))

print("\n\n❄️  TOP 10 NÚMEROS MAIS FRIOS:")
print(quentes_frios.tail(10).to_string(index=False))

# Visualização
fig, ax = plt.subplots(figsize=(14, 6))

colors = ['red' if x > 0 else 'blue' for x in quentes_frios['desvio_percentual']]
bars = ax.barh(quentes_frios['numero'].astype(str), quentes_frios['desvio_percentual'], color=colors, alpha=0.7)

ax.axvline(0, color='black', linestyle='--', linewidth=1)
ax.set_xlabel('Desvio Percentual (%)', fontsize=12, fontweight='bold')
ax.set_ylabel('Número', fontsize=12, fontweight='bold')
ax.set_title('Números Quentes (vermelho) vs Frios (azul)', fontsize=14, fontweight='bold')
ax.grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.show()
```
